```python
import numpy as np
from tabulate import tabulate
import copy
import pandas as pd
```

```python
matrix = []

with open("Data", "r") as file:
    for line in file:
        row = [int(num.strip()) for num in line.strip().split(",") if num.strip()]
        matrix.append(row)

cell_num_for_tabulate = [i for i in range(len(matrix))]

print('Исходная таблица')
print(tabulate(matrix, headers=cell_num_for_tabulate, showindex=cell_num_for_tabulate, tablefmt="grid"))
```

```python
x = 4
y = 5
seats = np.zeros((x, y))

cell_num_for_tabulate_seats_vert = [i for i in range(len(seats))]
cell_num_for_tabulate_seats_hor = [i for i in range(len(seats[0]))]

print('Пустая карта расстановки')
print(tabulate(seats, headers=cell_num_for_tabulate_seats_hor, showindex=cell_num_for_tabulate_seats_vert, tablefmt="grid"))
```

```python
selected = []
size_selected = 0
size_matrix = len(matrix)

iter = 0
while size_selected < size_matrix:
    iter += 1
    print('Текущие выбранные элементы')
    print(selected)
    print(f'Шаг {iter}')
    if len(selected) == 0:
        sums = [sum(i) for i in matrix]
        print('Суммы')
        print(sums)
        max_sum = max(sums)
        print(f'Максимальная сумма {max_sum}')
        print('Добавлено в размещение ', end='')
        for num, item in enumerate(sums):
            if item == max_sum:
                print(num, ' ', end='')
                selected.append(num)
        print()
    else:
        sums = []
        for num_row, row in enumerate(matrix):
            cur_sum = 0
            for num_col, item in enumerate(matrix[num_row]):
                if num_col in selected:
                    cur_sum += item
                else:
                    cur_sum -= item
            sums.append(cur_sum)
        max_sum = -np.inf
        for num, item in enumerate(sums):
            if item > max_sum and num not in selected:
                max_sum = item
        print('Суммы')
        print(sums)
        print(f'Максимальная сумма {max_sum}')
        print('Добавлено в размещение ', end='')
        for num, item in enumerate(sums):
            if item == max_sum and num not in selected:
                print(num, ' ', end='')
                selected.append(num)
        print()
    print('Hа выходе итерации следующие выбранные элементы')
    print(selected)
    print('\n\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\n\n\n\n\n')
    size_selected = len(selected)
```

```python
for i in range(-1, max(len(seats), len(seats[0]))):
    print(f'Текущая позиция заполнения {i}')
    print('Оставшиеся элементы для заполнения: ')
    print(selected)
    for j in range(i+1):
        if i + 1 < len(seats[0]) and j < len(seats):
            seats[j][i+1] = selected[0] if selected else -1
            selected.pop(0)
    for j in range(i+1):
        if j < len(seats[0]) and i + 1 < len(seats):
            seats[i+1][j] = selected[0] if selected else -1
            selected.pop(0)
    if i + 1 < len(seats[0]) and i + 1 < len(seats):
        seats[i+1][i+1] = selected[0] if selected else -1
        selected.pop(0)
    print('После итерации')
    print(tabulate(seats, headers=cell_num_for_tabulate_seats_hor, showindex=cell_num_for_tabulate_seats_vert, tablefmt="grid"))
    print('\n\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\n\n\n\n\n')

    

print('Текущая карта расстановки на первом этапе')
print(tabulate(seats, headers=cell_num_for_tabulate_seats_hor, showindex=cell_num_for_tabulate_seats_vert, tablefmt="grid"))
```

```python
def find_index(matrix, target):
    for i, row in enumerate(matrix):
        for j, value in enumerate(row):
            if value == target:
                return (i, j)
    return None

def all_weight(seats, matrix):
    w = 0
    for i in range(len(matrix)):
        (x_1, y_1) = find_index(seats, i)
        for j in range(i+1, len(matrix)):
            if matrix[i][j] != 0:
                (x_2, y_2) = find_index(seats, j)
                w += matrix[i][j] * (abs(x_1 - x_2) + abs(y_1 - y_2))
    return w

def conc_weight(seats, matrix, num):
    w = 0
    l = 0
    (x_1, y_1) = find_index(seats, num)
    for j in range(0, len(matrix)):
        if matrix[num][j] != 0:
            (x_2, y_2) = find_index(seats, j)
            w += matrix[num][j] * (abs(x_1 - x_2) + abs(y_1 - y_2))
            l += matrix[num][j]
    return w/l

def x_weight(seats, matrix, num):
    w = 0
    l = 0
    (x_1, y_1) = find_index(seats, num)
    for j in range(0, len(matrix)):
        if matrix[num][j] != 0:
            (x_2, y_2) = find_index(seats, j)
            w += matrix[num][j] * (abs(x_1 - x_2))
            l += matrix[num][j]
    return w/l

def y_weight(seats, matrix, num):
    w = 0
    l = 0
    (x_1, y_1) = find_index(seats, num)
    for j in range(0, len(matrix)):
        if matrix[num][j] != 0:
            (x_2, y_2) = find_index(seats, j)
            w += matrix[num][j] * (abs(y_1 - y_2))
            l += matrix[num][j]
    return w/l

def replace_weight(seats, matrix, x1, y1, x2, y2):
    tmp_seats = copy.deepcopy(seats)
    tmp = tmp_seats[x1][y1]
    tmp_seats[x1][y1] = tmp_seats[x2][y2]
    tmp_seats[x2][y2] = tmp
    return all_weight(tmp_seats, matrix)
    
```

```python
old_weight = all_weight(seats, matrix)
old_seats = copy.deepcopy(seats)
print(f'Текущаяя сумма весов: {old_weight}')    
print(tabulate(old_seats, headers=cell_num_for_tabulate_seats_hor, showindex=cell_num_for_tabulate_seats_vert, tablefmt="grid"))
```

```python
prev_weight = 0
prev_prev_weight = 0
cur_weight = all_weight(seats, matrix)
iter = 0
while prev_prev_weight != cur_weight or cur_weight != min(cur_weight, prev_weight):
    print(f'Текущая итерация: {iter}')
    print(tabulate(seats, headers=cell_num_for_tabulate_seats_hor, showindex=cell_num_for_tabulate_seats_vert, tablefmt="grid"))

    weights = [conc_weight(seats, matrix, i) for i in range(len(matrix))]
    max_weight = max(weights)
    max_index = weights.index(max_weight)
    (v_x, v_y) = find_index(seats, max_index)
    
    print('Распределение средних весов')
    print([f'{i:.4f}' for i in weights])
    print(f'Максимальный вес {max_weight:.4f} у элемента с индексом {max_index}')
    

    xw = x_weight(seats, matrix, max_index) - 1
    yw = y_weight(seats, matrix, max_index) - 1
    print(f'Центр масс для этого элемента: ({xw:.3f},{yw:.3f})\n')
    
    x_min = 0
    y_min = 0
    min_w = np.inf
    
    for i in range(2 if xw != int(xw) else 1):
        for j in range(2 if yw != int(yw) else 1):
            x_c = int(xw) + i
            y_c = int(yw) + j
            print(f'Проверяем с {x_c}, {y_c} - {seats[x_c][y_c]}')
            cur_w = replace_weight(seats, matrix, v_x, v_y, x_c, y_c)
            if cur_w < min_w:
                print(f'Пока нашли минимум с перестановкой')
                x_min = x_c
                y_min = y_c
                min_w = cur_w 
    print('Проверка закончена')
    if min_w != np.inf:
        print(f'Выполняем перестановку {seats[x_min][y_min]} - {seats[v_x][v_y]}')
        tmp = seats[x_min][y_min]
        seats[x_min][y_min] = seats[v_x][v_y]
        seats[v_x][v_y] = tmp

        
    print('После итерации получается следующая расстановка')
    print(tabulate(seats, headers=cell_num_for_tabulate_seats_hor, showindex=cell_num_for_tabulate_seats_vert, tablefmt="grid"))
    print(f'Сумма весов {all_weight(seats, matrix)}')
    iter += 1
    prev_prev_weight = prev_weight
    prev_weight = cur_weight
    cur_weight = all_weight(seats, matrix)
    print('\n\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\n\n\n\n\n')

        
    
    
```

```python
pot_seats = copy.deepcopy(seats)
k = 0
for i in range(len(seats)):
    for j in range(len(seats[i])):
        pot_seats[i][j]=k
        k+=1
pot_weight = all_weight(pot_seats, matrix)
        
print(f'Для примера последовательное расзмещение с длиной {pot_weight}')
print(tabulate(pot_seats, headers=cell_num_for_tabulate_seats_hor, showindex=cell_num_for_tabulate_seats_vert, tablefmt="grid"))


print(f'До итерационного этапа длина проводников оценивалась как {old_weight}')
print(tabulate(old_seats, headers=cell_num_for_tabulate_seats_hor, showindex=cell_num_for_tabulate_seats_vert, tablefmt="grid"))


print(f'После итерационного этапа получается следующая расстановка элементов с длинами проводников {all_weight(seats, matrix)}')
print(tabulate(seats, headers=cell_num_for_tabulate_seats_hor, showindex=cell_num_for_tabulate_seats_vert, tablefmt="grid"))
```
