```python
import numpy as np
import copy
from tabulate import tabulate
```

```python
matrix = []

with open("Data.txt", "r") as file:
    for line in file:
        row = [int(num.strip()) for num in line.strip().split(",") if num.strip()]
        matrix.append(row)

cell_num_for_tabulate = [i for i in range(len(matrix))]

print('Исходная таблица')
print(tabulate(matrix, headers=cell_num_for_tabulate, showindex=cell_num_for_tabulate, tablefmt="grid"))
    
```

```python


group_count = [4, 5, 5, 6]
grouped_indices = []
group_of_indicies = []
matrix_size = len(matrix)

last_ended_group = 0



while len(grouped_indices) < matrix_size:
    print('------------------------------------------')

    dict_min = {
        'value' : np.inf,
        'count' : 0,
        'index' : -1,
        'indices' : []
    }
    dict_of_all = {
        'indices' : [],
        'sums' : []
    }
    print('Распределенные: ', grouped_indices)
    for i, row in enumerate(matrix):
        if i not in grouped_indices:
            row_sum = sum(cell for cell_num, cell in enumerate(row) if cell_num not in grouped_indices)
            non_zero_count = sum(1 for cell_num, cell in enumerate(row) if cell_num not in grouped_indices and cell != 0)
            non_zero_indices = [cell_num for cell_num, cell in enumerate(row) if cell_num not in grouped_indices and cell != 0]
            
            dict_of_all['indices'].append(i)
            dict_of_all['sums'].append(row_sum)
            
            if row_sum == dict_min['value']:
                if non_zero_count > dict_min['count'] - 1:
                    dict_min['count'] = non_zero_count + 1
                    dict_min['index'] = i
                    dict_min['indices'] = non_zero_indices + [i]
            elif row_sum < dict_min['value']:
                dict_min['value'] = row_sum
                dict_min['count'] = non_zero_count + 1
                dict_min['index'] = i
                dict_min['indices'] = non_zero_indices + [i]
    print(tabulate([dict_of_all['indices'],dict_of_all['sums']], showindex=['index', 'sum'], tablefmt="grid"))
    print(dict_min)

    grouped_indices_count = sum(1 for item in group_of_indicies if item == last_ended_group)
    print("занято: ", grouped_indices_count, "   хотят ", dict_min['count'], "   доступно ", group_count[last_ended_group])
    if grouped_indices_count + dict_min['count'] <= group_count[last_ended_group]:
        for item in dict_min['indices']:
            grouped_indices.append(item)
            group_of_indicies.append(last_ended_group)
        print('Просто добавили. Теперь так: ', grouped_indices, '\n', 'И так ', group_of_indicies)
    else:
        print('Просто не получилось. Будет сложно')
        
        while(grouped_indices_count + dict_min['count'] > group_count[last_ended_group]):
            print('------ -------- ------')
            
            print('Размер оставшихся ', dict_min['count'], '  Оставшиеся ', dict_min['indices'])
            
            marked = dict_min['indices']
            print('Помеченные: ', marked)
            delta_array = []
            delta_array_indices = []
            for pos, ind in enumerate(marked):
                print('-------')
                print("Это все помеченные свободные: ", [[i, val] for i, val in enumerate(matrix[ind]) if i in marked])
                print('Считается для ', ind)
                
                delta = dict_of_all['sums'][pos] - sum(val for i, val in enumerate(matrix[ind]) if i in marked and i)
                print(f'Из {dict_of_all['sums'][pos]} вычитается {sum(val for i, val in enumerate(matrix[ind]) if i in marked)} и получается {delta}')
                delta_array.append(delta)
                delta_array_indices.append(ind)
            print('Список идексов в дельте: ', delta_array_indices)
            delta_min = min(delta_array)
            
            delta_min_position = delta_array.index(delta_min)
            print('Что удаляется: ', delta_min_position)
            dict_min['indices'].pop(delta_min_position)
            dict_min['count'] -= 1
        for item in dict_min['indices']:
            grouped_indices.append(item)
            group_of_indicies.append(last_ended_group)
        print('Сложно добавили. Теперь так: ', grouped_indices, '\n', 'И так ', group_of_indicies)

    grouped_indices_count = sum(1 for item in group_of_indicies if item == last_ended_group)
    if grouped_indices_count >= group_count[last_ended_group]:
        last_ended_group += 1
            
print('Итого у нас полчается такое распределение по узлам (сверху) и по группам (снизу)')
print(group_of_indicies)
print(grouped_indices)
```

```python
def get_group(groups, item):
    for num, group in enumerate(groups):
        if item in group:
            return num
```

```python
def print_matrix(matrix):
    for i in range(len(matrix)):
        for j in range(len(matrix[i])):
            print()
```

```python
getted_cells = 0



for iter in range(len(group_count)-1):
    print()
    groups = []
    for i in range(len(group_count)):
        groups.append([grouped_indices[j] for j in range(len(group_of_indicies)) if group_of_indicies[j] == i])
    
    matrix_prev = []
    matrix_cur = copy.deepcopy(groups)
    iter_iter = 0
    while matrix_cur != matrix_prev:
        iter_iter += 1
        print('-------------------------------------------', end='\n\n\n')
        print('Итерация №', iter+1, '.', iter_iter)
        horizontal = 0
        horizontal_indices = []
        vertical = group_count[iter]
        vertical_indices = matrix_cur[iter]
        for i in range(iter + 1, len(group_count)):
            horizontal += group_count[i]
            horizontal_indices += matrix_cur[i]
        matrix_iter = np.zeros((vertical, horizontal))
        
        for i in range(vertical):
            for j in range(horizontal):
                hor_index = horizontal_indices[j]
                ver_index = vertical_indices[i]
                hor_group = get_group(matrix_cur, hor_index)
                ver_group = get_group(matrix_cur, ver_index)
                
                sum_in_vert_row_hor_group = sum(matrix[ver_index][k] for k in range(matrix_size) if k in matrix_cur[hor_group])
                sum_in_vert_row_vert_group = sum(matrix[ver_index][k] for k in range(matrix_size) if k in matrix_cur[ver_group])
                sum_in_hor_row_hor_group = sum(matrix[hor_index][k] for k in range(matrix_size) if k in matrix_cur[hor_group])
                sum_in_hor_row_vert_group = sum(matrix[hor_index][k] for k in range(matrix_size) if k in matrix_cur[ver_group])
                
                z = (sum_in_vert_row_hor_group - sum_in_vert_row_vert_group) + (sum_in_hor_row_vert_group - sum_in_hor_row_hor_group) - 2 * matrix[ver_index][hor_index]
                
                matrix_iter[i][j] = z
        print('Таблица dR')
        print(tabulate(matrix_iter, headers=horizontal_indices, showindex=vertical_indices, tablefmt="grid"))
        
        positive_indices = [(i, j) for i, row in enumerate(matrix_iter) for j, value in enumerate(row) if value > 0]
        if positive_indices:
            max_positive = max(matrix_iter[i][j] for i, j in positive_indices)
        else:
            print('Нет положительных значений (>0)')
            break
        
        for i in range(vertical):
            for j in range(horizontal):
                if max_positive == matrix_iter[i][j]:
                    max_i = i
                    max_j = j
                
        matrix_prev = copy.deepcopy(matrix_cur)
        
        tmp_group_ver = get_group(matrix_cur, vertical_indices[max_i])
        tmp_group_hor = get_group(matrix_cur, horizontal_indices[max_j])
        
        tmp_pos_ver = max_i
        tmp_pos_hor = max_j - sum(group_count[i] for i in range(iter + 1, tmp_group_hor))
        

        print('макcимальное значение: ', max_positive)
        print('Находится в ячейке', max_i, max_j, ' с названием ', vertical_indices[max_i], horizontal_indices[max_j])
        
        print('Нужно заменить индекс ', vertical_indices[max_i], ' в группе ', get_group(matrix_cur, vertical_indices[max_i]), ' который стоит на позиции ', tmp_pos_ver)
        print('На индекс ', horizontal_indices[max_j], ' в группе ', get_group(matrix_cur, horizontal_indices[max_j]), ' который стоит на позиции ', tmp_pos_hor, 'вот группа')
        print(tabulate(matrix_cur, tablefmt="grid"))
        
        tmp_1 = matrix_cur[tmp_group_ver][tmp_pos_ver]
        tmp_2 = matrix_cur[tmp_group_hor][tmp_pos_hor]
        matrix_cur[tmp_group_ver][tmp_pos_ver] = tmp_2
        matrix_cur[tmp_group_hor][tmp_pos_hor] = tmp_1
        
        print('\n\n\n------|| На выходе из итерации получается следующее распределение ||------')
        print(tabulate(matrix_cur, tablefmt="grid"))
        print('Конец итерации ', iter, '.', iter_iter)
        print('\n\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\\/\n')
```
